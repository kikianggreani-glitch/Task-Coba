```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv('/content/sample_data/Smartphone_Usage_Productivity_Dataset_50000.csv')
df.head()
```

```python
df.columns
```

```python
kategori = df['Occupation'].value_counts().index
jumlah = df['Occupation'].value_counts().values

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(kategori, jumlah, color='pink', alpha=0.5)

plt.grid(linestyle='-', linewidth=2, axis='y', alpha=0.5)

for i in range(len(jumlah)):
    plt.text(i, jumlah[i], jumlah[i], ha='center', va='bottom')

plt.xlabel('Occupation')
plt.ylabel('Jumlah')
plt.title('Jumlah Pengguna Berdasarkan Occupation')

plt.xticks(rotation=45)
plt.show()
```

```python
avg_data = df.groupby('Daily_Phone_Hours')['Stress_Level'].mean()

x = avg_data.index
y = avg_data.values

plt.plot(x, y, 'b*-')
plt.xlabel("Daily Phone Hours")
plt.ylabel("Average Stress Level")
plt.title('Stress Level berdasarkan Phone Usage')

plt.show()
```

```python
import matplotlib.pyplot as plt

# menghitung rata-rata sleep hours berdasarkan penggunaan HP
avg_data = df.groupby('Daily_Phone_Hours')['Sleep_Hours'].mean()

x = avg_data.index
y = avg_data.values

plt.plot(x,y)
plt.xlabel("Daily Phone Hours")
plt.ylabel("Average Sleep Hours")
plt.title('Hubungan Penggunaan HP dengan Waktu Tidur')

plt.show()
```

```python
import matplotlib.pyplot as plt

# menghitung jumlah data berdasarkan usia
age_count = df['Age'].value_counts().sort_index()

x = age_count.index
y = age_count.values

plt.plot(x,y)
plt.xlabel("Age")
plt.ylabel("Jumlah Pengguna")
plt.title('Jumlah Pengguna Berdasarkan Usia')

plt.show()
```

```python
import matplotlib.pyplot as plt

gender_count = df['Gender'].value_counts()

labels = gender_count.index
sizes = gender_count.values

plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title('Persentase Gender Pengguna')

plt.show()
```
